```python
import pandas as pd
import json
df = pd.read_parquet("hf://datasets/dair-ai/emotion/unsplit/train-00000-of-00001.parquet")
# define label dictionary mapping integers ot emotions
label_dict = {0:"sadness", 1:"joy", 2:"love", 3:"anger", 4:"fear", 5:"surprise" }
# create a subset of tweets 
tweets = [{"id":r[0], "text":r[1].text, "cat":label_dict[r[1].label]} for r in df[0:200].iterrows()]
# save tweets as JSON
with open("tweets.json", "w", encoding="utf-8") as f:
    json.dump(tweets, f, indent=4)
```
